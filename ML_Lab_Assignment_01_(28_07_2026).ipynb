# LIST

Q1. Create a list of 10 integers. Write a Python program to find the largest element, smallest element, sum, and average of the list.

```python
import random

# Create a list of 10 random integers between 1 and 100
numbers = [random.randint(1, 100) for _ in range(10)]

print(f"The list of numbers is: {numbers}")

# Find the largest element
largest_element = max(numbers)
print(f"Largest element: {largest_element}")

# Find the smallest element
smallest_element = min(numbers)
print(f"Smallest element: {smallest_element}")

# Calculate the sum
total_sum = sum(numbers)
print(f"Sum of elements: {total_sum}")

# Calculate the average
average = total_sum / len(numbers)
print(f"Average of elements: {average}")
```

Q2. Given a list of integers, insert a new element at a specified position, delete an element, and reverse the list without using the built-in reverse() method

```python
# Insert a new element (e.g., 99) at a specified position (e.g., index 2)
new_element = 99
insert_position = 2

print(f"Original list: {numbers}")
numbers.insert(insert_position, new_element)
print(f"List after inserting {new_element} at index {insert_position}: {numbers}")
```

```python
# Delete an element (e.g., the first occurrence of 97)
# You can also delete by index using `del numbers[index]` or `numbers.pop(index)`
element_to_delete = 97

print(f"List before deleting {element_to_delete}: {numbers}")
if element_to_delete in numbers:
    numbers.remove(element_to_delete)
    print(f"List after deleting {element_to_delete}: {numbers}")
else:
    print(f"{element_to_delete} not found in the list.")
```

```python
# Reverse the list using slicing [::-1]
print(f"Original list: {numbers}")
reversed_numbers = numbers[::-1]
print(f"Reversed list (using slicing): {reversed_numbers}")
```

# NESTED LIST

Q3. Create a 3 × 3 nested list (matrix) and write a program to print the second row, third column, and the sum of all elements.

```python
# Create a 3x3 nested list (matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(f"The matrix is:\n{matrix}")

# Print the second row (index 1)
second_row = matrix[1]
print(f"Second row: {second_row}")

# Print the third column (index 2)
third_column = [row[2] for row in matrix]
print(f"Third column: {third_column}")

# Calculate the sum of all elements
total_sum = 0
for row in matrix:
    for element in row:
        total_sum += element
print(f"Sum of all elements: {total_sum}")
```

Q4. Given a nested list containing marks of students in three subjects, calculate the average marks of each student and identify the student with the highest average.

```python
# Create a nested list of student marks (Student Name, Subject1, Subject2, Subject3)
student_marks = [
    ['Alice', 85, 90, 78],
    ['Bob', 75, 88, 92],
    ['Charlie', 90, 85, 80],
    ['David', 60, 70, 65]
]

print("Student Marks:")
for student in student_marks:
    print(student)

# Initialize variables to keep track of highest average
highest_average = -1
student_with_highest_avg = ''

print("\nCalculating Averages:")
for student_data in student_marks:
    student_name = student_data[0]
    marks = student_data[1:] # Slice to get only the marks

    # Calculate the average for the current student
    average_marks = sum(marks) / len(marks)
    print(f"{student_name}'s average marks: {average_marks:.2f}")

    # Check if this student has the highest average so far
    if average_marks > highest_average:
        highest_average = average_marks
        student_with_highest_avg = student_name

print(f"\nStudent with the highest average marks: {student_with_highest_avg} (Average: {highest_average:.2f})")
```

Q5. Create a dictionary to store a student's name, roll number, department, and CGPA. Display all the details and update the CGPA.

Q6. Given a dictionary containing subject names and marks, write a program to display all keys, all values, and the subject with the highest marks.

```python
# Create a dictionary of subject names and marks
subject_marks = {
    'Math': 95,
    'Science': 88,
    'History': 75,
    'English': 92,
    'Art': 80
}

print("Subject Marks Dictionary:")
print(subject_marks)

# Display all keys (subject names)
print("\nAll Subject Names (Keys):")
for subject in subject_marks.keys():
    print(subject)

# Display all values (marks)
print("\nAll Marks (Values):")
for marks in subject_marks.values():
    print(marks)

# Find the subject with the highest marks
highest_marks = 0
highest_scoring_subject = ''

for subject, marks in subject_marks.items():
    if marks > highest_marks:
        highest_marks = marks
        highest_scoring_subject = subject

print(f"\nSubject with the highest marks: {highest_scoring_subject} (Marks: {highest_marks})")
```

Q7. Create a dictionary where each key represents a department and the corresponding value is a list of employee names. Display all employees of a given department and count the employees in each department.

```python
# Create a dictionary where keys are departments and values are lists of employee names
departments = {
    'HR': ['Alice', 'Bob', 'Charlie'],
    'Engineering': ['David', 'Eve', 'Frank', 'Grace'],
    'Sales': ['Heidi', 'Ivan']
}

print("Department and Employee List:")
for dept, employees in departments.items():
    print(f"{dept}: {', '.join(employees)}")

# Display all employees of a given department
given_department = 'Engineering'
if given_department in departments:
    print(f"\nEmployees in {given_department}: {', '.join(departments[given_department])}")
else:
    print(f"\nDepartment '{given_department}' not found.")

# Count the employees in each department
print("\nEmployee Count per Department:")
for dept, employees in departments.items():
    print(f"{dept}: {len(employees)} employees")
```

Q9. Create a list of dictionaries where each dictionary stores a student's name and marks. Write a program to display all student names and identify the topper.

```python
# Create a list of dictionaries, each representing a student
students = [
    {'name': 'Alice', 'marks': 95},
    {'name': 'Bob', 'marks': 88},
    {'name': 'Charlie', 'marks': 92},
    {'name': 'David', 'marks': 98},
    {'name': 'Eve', 'marks': 85}
]

print("All Student Names:")
for student in students:
    print(student['name'])

# Identify the topper
topper_name = ''
highest_marks = -1

for student in students:
    if student['marks'] > highest_marks:
        highest_marks = student['marks']
        topper_name = student['name']

print(f"\nThe topper is: {topper_name} with {highest_marks} marks.")
```

Q10. Create a list of dictionaries containing book titles and prices. Write a program to find the most expensive book and calculate the total cost of all books.

```python
# Create a list of dictionaries containing book titles and prices
books = [
    {'title': 'The Hitchhiker\'s Guide to the Galaxy', 'price': 12.50},
    {'title': 'Pride and Prejudice', 'price': 8.75},
    {'title': '1984', 'price': 15.00},
    {'title': 'To Kill a Mockingbird', 'price': 10.25},
    {'title': 'The Great Gatsby', 'price': 11.99}
]

print("List of Books:")
for book in books:
    print(f"  {book['title']}: ${book['price']:.2f}")

# Find the most expensive book
most_expensive_book = None
highest_price = -1

for book in books:
    if book['price'] > highest_price:
        highest_price = book['price']
        most_expensive_book = book

if most_expensive_book:
    print(f"\nMost expensive book: {most_expensive_book['title']} (${most_expensive_book['price']:.2f})")
else:
    print("\nNo books found.")

# Calculate the total cost of all books
total_cost = 0
for book in books:
    total_cost += book['price']

print(f"Total cost of all books: ${total_cost:.2f}")
```

Q12. Given an array of integers, insert a new element, delete an existing element, and display the array in reverse order.

```python
# Using the array_elements from the previous question (Q11)
# Make a copy to avoid modifying the original list if it's used elsewhere
my_array = list(array_elements)

print(f"Original array: {my_array}")

# 1. Insert a new element (e.g., 500) at a specified position (e.g., index 3)
new_element_to_insert = 500
insert_position_array = 3
my_array.insert(insert_position_array, new_element_to_insert)
print(f"Array after inserting {new_element_to_insert} at index {insert_position_array}: {my_array}")

# 2. Delete an existing element (e.g., the first occurrence of 83 if it exists)
# We'll pick a value that is likely to be in the array or handle if it's not.
# Let's try to delete the element at index 0 for a consistent example
element_to_delete_value = my_array[0] # Get the first element to delete it

print(f"\nArray before deleting {element_to_delete_value}: {my_array}")
if element_to_delete_value in my_array:
    my_array.remove(element_to_delete_value)
    print(f"Array after deleting {element_to_delete_value}: {my_array}")
else:
    print(f"{element_to_delete_value} not found in the array.")

# 3. Display the array in reverse order using slicing
print(f"\nOriginal array: {my_array}")
reversed_array = my_array[::-1]
print(f"Reversed array (using slicing): {reversed_array}")
```

Q14. Create a tuple containing student details (name, age, department, CGPA). Convert the tuple into a list, update the CGPA, and convert it back into a tuple.

# SET

Q15. Create two sets of integers and write a program to find their union, intersection, difference, and symmetric difference.

```python
# Create two sets of integers
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(f"Set 1: {set1}")
print(f"Set 2: {set2}")

# Find the union of the two sets
union_set = set1.union(set2)
print(f"\nUnion of Set 1 and Set 2: {union_set}")

# Find the intersection of the two sets
intersection_set = set1.intersection(set2)
print(f"Intersection of Set 1 and Set 2: {intersection_set}")

# Find the difference (elements in set1 but not in set2)
difference_set1_set2 = set1.difference(set2)
print(f"Difference (Set 1 - Set 2): {difference_set1_set2}")

# Find the difference (elements in set2 but not in set1)
difference_set2_set1 = set2.difference(set1)
print(f"Difference (Set 2 - Set 1): {difference_set2_set1}")

# Find the symmetric difference (elements in either set, but not in both)
symmetric_difference_set = set1.symmetric_difference(set2)
print(f"Symmetric Difference: {symmetric_difference_set}")
```

Q16. Given a list containing duplicate elements, use a set to remove duplicates and display the unique elements along with their count.

```python
# Create a list containing duplicate elements
original_list_with_duplicates = [1, 2, 2, 3, 4, 4, 4, 5, 6, 1, 7, 8, 8]
print(f"Original list with duplicates: {original_list_with_duplicates}")

# Use a set to remove duplicates
unique_elements_set = set(original_list_with_duplicates)

# Convert the set back to a list to display ordered unique elements (optional)
unique_elements_list = list(unique_elements_set)
print(f"\nUnique elements (as a list): {unique_elements_list}")

# Display the count of unique elements
count_unique_elements = len(unique_elements_set)
print(f"Count of unique elements: {count_unique_elements}")
```
