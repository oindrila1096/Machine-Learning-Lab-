```python
import pandas as pd

# Create a Pandas Series containing numbers from 1 to 10
series = pd.Series(range(1, 11))

# Convert Series into a DataFrame
df = series.to_frame(name = "Numbers")

# Print Series
print("Series :- ")
print(series)

# Print DataFrame
print("\nDataFrame :- ")
print(df)

# Print Shape
print("\nShape :- " , df.shape)

# Print Datatype
print("\nDatatype :- " , series.dtype)

# Print Sum of elements
print("\nSum of elements :- " , series.sum())

# Print Mean of elements
print("\nMean of elements :- " , series.mean())

# Print Maximum value
print("\nMaximum value :- " , series.max())

# Print Minimum value
print("\nMinimum value :- " , series.min())
```

```python
import pandas as pd

# Create DataFrame A
A = pd.DataFrame([[1,2,3], [4,5,6], [7,8,9]],
                 columns = ["C1", "C2", "C3"], index = ["R1", "R2", "R3"])

# Create DataFrame B
B = pd.DataFrame([[9,8,7], [6,5,4], [3,2,1]],
                 columns = ["C1", "C2", "C3"], index = ["R1", "R2", "R3"])

# Print DataFrames
print("DataFrame A :- ")
print(A)

print("\nDataFrame B :- ")
print(B)

# DataFrame Addition
print("\nDataFrame Addition :- ")
print(A + B)

# DataFrame Subtraction
print("\nDataFrame Subtraction :- ")
print(A - B)

# Element-wise Multiplication
print("\nElement-wise Multiplication :- ")
print(A * B)

# Matrix Multiplication
print("\nMatrix Multiplication :- ")
print(A.to_numpy().dot(B.to_numpy()))
```

```python
import pandas as pd

# Create DataFrame
data = {"Student" : ["Amit", "Bhavna", "Chirag", "Divya", "Esha"],
        "Maths" : [85, 70, 95, 60, 88], "Science" : [90, 65, 91, 75, 84],
        "English" : [88, 72, 93, 70, 86], "History" : [92, 68, 96, 80, 90]}
df = pd.DataFrame(data)
df.set_index("Student", inplace = True)

print("Student Marks :- ")
print(df)

# Average marks of each student
print("\nAverage marks of each student :- ")
print(df.mean(axis = 1))

# Average marks of each subject
print("\nAverage marks of each subject :- ")
print(df.mean(axis = 0))

# Highest and lowest marks
print("\nHighest marks :- " , df.max().max())
print("Lowest marks :- " , df.min().min())

# Student having the highest total marks
df["Total"] = df.sum(axis = 1)
print("\nStudent having the highest total marks :- " ,
      df[df["Total"] == df["Total"].max()].index[0])

# Sort DataFrame by total marks in descending order
print("\nSorted DataFrame by total marks :- ")
print(df.sort_values(by = "Total", ascending = False))
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    "Day" : ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"],
    "Temperature" : [30, np.nan, 32, 29, np.nan, 35, 31]})

print("Original DataFrame :- ")
print(df)

# Identify missing values
print("\nMissing values :- ")
print(df.isnull())

# Fill missing values with mean temperature
mean_temp = df["Temperature"].mean()
df_mean = df.copy()
df_mean["Temperature"] = df_mean["Temperature"].fillna(mean_temp)

print("\nDataFrame after filling missing values with mean temperature :- ")
print(df_mean)

# Fill missing values using forward fill
df_ffill = df.copy()
df_ffill["Temperature"] = df_ffill["Temperature"].ffill()

print("\nDataFrame after filling missing values using forward fill :- ")
print(df_ffill)

# Drop rows containing missing values
df_drop = df.dropna()

print("\nDataFrame after dropping rows containing missing values :- ")
print(df_drop)

# Calculate average temperature after cleaning
print("\nAverage temperature after cleaning :- " ,
      df_drop["Temperature"].mean())
```

```python
import pandas as pd

# Create DataFrame
data = {"Product" : ["Product A", "Product B", "Product C", "Product D"],
        "Jan" : [120, 100, 150, 90], "Feb" : [130, 110, 145, 100],
        "Mar" : [140, 120, 160, 110], "Apr" : [150, 130, 170, 120],
        "May" : [160, 140, 180, 130], "Jun" : [170, 150, 190, 140]}
df = pd.DataFrame(data)

print("Sales Data :- ")
print(df)

# Total sales of each product
df["Total"] = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].sum(axis = 1)

print("\nTotal sales of each product :- ")
print(df[["Product", "Total"]])

# Average monthly sales of each product
df["Average"] = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].mean(axis = 1)

print("\nAverage monthly sales of each product :- ")
print(df[["Product", "Average"]])

# Best-selling product overall
best_product = df.loc[df["Total"].idxmax(), "Product"]

print("\nBest-selling product overall :- ", best_product)

# Month with highest overall sales
monthly_sales = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].sum()
highest_month = monthly_sales.idxmax()

print("\nMonth with highest overall sales :- ", highest_month)
print("Highest overall sales :- ", monthly_sales.max())

# Standard deviation of sales for each product
df["Standard Deviation"] = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].std(axis=1)

print("\nStandard deviation of sales for each product :- ")
print(df[["Product", "Standard Deviation"]])

# Create Quarter column
quarter_data = []

for i, row in df.iterrows():
    quarter_data.append({
        "Product": row["Product"],
        "Q1": row["Jan"] + row["Feb"] + row["Mar"],
        "Q2": row["Apr"] + row["May"] + row["Jun"]
    })

quarter_df = pd.DataFrame(quarter_data)

print("\nTotal sales per product per quarter :- ")
print(quarter_df)
```
