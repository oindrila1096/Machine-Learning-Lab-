```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create a graph representing the friendship network
G = nx.Graph()

friendships = [
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "David"),
    ("David", "Eva"),
    ("Eva", "Frank"),
    ("Charlie", "Frank")
]

G.add_edges_from(friendships)

# 2. Visualize the network graph
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(8, 6))

nx.draw_networkx_nodes(
    G,
    pos,
    node_color="lightblue",
    node_size=2000
)

nx.draw_networkx_edges(
    G,
    pos,
    edge_color="gray",
    width=2
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

plt.title("Student Friendship Network")
plt.axis("off")
plt.show()
print()

# 3. Display total number of nodes and edges
print("Total number of nodes:", G.number_of_nodes())
print("Total number of edges:", G.number_of_edges())

# 4. Compute and print degree of each student
print("\nDegree of each student:")
for student, degree in G.degree():
    print(student, ":", degree)

# 5. Identify student(s) with the highest number of friends
degrees = dict(G.degree())
highest_degree = max(degrees.values())

highest_degree_students = [
    student for student, degree in degrees.items()
    if degree == highest_degree
]

print("\nStudent(s) with the highest number of friends:",
      highest_degree_students)
print("Highest number of friends:", highest_degree)

# 6. Check direct friendship between Alice and Eva
print("\nDirect friendship between Alice and Eva:",
      "Yes" if G.has_edge("Alice", "Eva") else "No")

# 7. Find shortest path between Alice and Frank
shortest_path = nx.shortest_path(G, "Alice", "Frank")
path_length = nx.shortest_path_length(G, "Alice", "Frank")

print("\nShortest path from Alice to Frank:", shortest_path)
print("Length of shortest path:", path_length)

# 8. Observations
print("\nObservations:")
print("1. Charlie and David have the highest number of friends, with 3 friends each.")
print("2. The friendship network is connected, and Alice can reach Frank through Charlie.")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create a graph representing the tower network
G = nx.Graph()

tower_connections = [
    ("T1", "T2"),
    ("T1", "T3"),
    ("T2", "T4"),
    ("T3", "T4"),
    ("T4", "T5"),
    ("T5", "T6"),
    ("T3", "T6"),
    ("T6", "T7"),
    ("T2", "T7")
]

G.add_edges_from(tower_connections)

# 2. Visualize the network graph
pos = nx.spring_layout(G, seed=42)

# Calculate betweenness centrality
betweenness = nx.betweenness_centrality(G)

# Find tower with highest betweenness centrality
highest_betweenness = max(betweenness.values())

critical_towers = [
    tower for tower, value in betweenness.items()
    if value == highest_betweenness
]

# Draw nodes
node_colors = [
    "red" if tower in critical_towers else "lightblue"
    for tower in G.nodes()
]

node_sizes = [
    2500 if tower in critical_towers else 1800
    for tower in G.nodes()
]

nx.draw_networkx_nodes(
    G,
    pos,
    node_color=node_colors,
    node_size=node_sizes
)

# Draw edges
nx.draw_networkx_edges(
    G,
    pos,
    edge_color="gray",
    width=2
)

# Draw labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

plt.title("Telecom Tower Network")
plt.axis("off")
plt.show()
print()

# 3. Calculate and display total number of nodes and edges
print("Total number of nodes:", G.number_of_nodes())
print("Total number of edges:", G.number_of_edges())

# 4. Compute degree of each tower
print("\nDegree of each tower:")
for tower, degree in G.degree():
    print(tower, ":", degree)

highest_degree = max(dict(G.degree()).values())

highest_degree_towers = [
    tower for tower, degree in G.degree()
    if degree == highest_degree
]

print("\nTower(s) with the highest degree:", highest_degree_towers)
print("Highest degree:", highest_degree)

# 5. Calculate shortest path between T1 and T7
shortest_path = nx.shortest_path(G, "T1", "T7")
path_length = nx.shortest_path_length(G, "T1", "T7")

print("\nShortest path from T1 to T7:", shortest_path)
print("Length of shortest path:", path_length)

# 6. Calculate centrality measures
degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
closeness_centrality = nx.closeness_centrality(G)

print("\nCentrality measures:")

for tower in G.nodes():
    print(
        tower,
        ": Degree Centrality =", round(degree_centrality[tower], 3),
        ", Betweenness Centrality =", round(betweenness_centrality[tower], 3),
        ", Closeness Centrality =", round(closeness_centrality[tower], 3)
    )

# 7. Check whether the network is fully connected
if nx.is_connected(G):
    print("\nNetwork is fully connected: Yes")
else:
    print("\nNetwork is fully connected: No")
    print("Number of connected components:", nx.number_connected_components(G))

# 8. Identify tower with highest betweenness centrality
print("\nTower(s) with the highest betweenness centrality:",
      critical_towers)
print("Highest betweenness centrality:",
      round(highest_betweenness, 3))

# 9. Observations
print("\nObservations:")
print("1. T3 and T6 have high connectivity and play important roles in the network.")
print("2. The network is fully connected, so every tower can communicate with every other tower.")
print("3. The tower with the highest betweenness centrality is the most critical for connecting different parts of the network.")
```
